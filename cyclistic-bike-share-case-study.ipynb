### PHASE 3 - PROCESS

#### **Pre-processing & Cleaning the data**
#### I noticed that the columns "started_at" & "ended_at" in April 2020 trip data was in character data type so let's change it to date-time format so that it matches other datasets when merging 

```r
trip_data_apr_2020$started_at <- as.POSIXct(trip_data_apr_2020$started_at)
trip_data_apr_2020$ended_at <- as.POSIXct(trip_data_apr_2020$ended_at)
```

#### Also noticed that the columns "start_station_id" & "end_station_id" in April 2020 to November 2020 trip datasets was in double data type so let's change it to character type to maintain compatibility

```r
trip_data_apr_2020 <- mutate(trip_data_apr_2020,start_station_id = as.character(start_station_id),end_station_id = as.character(end_station_id))
trip_data_may_2020 <- mutate(trip_data_may_2020,start_station_id = as.character(start_station_id),end_station_id = as.character(end_station_id))
trip_data_june_2020 <- mutate(trip_data_june_2020,start_station_id = as.character(start_station_id),end_station_id = as.character(end_station_id))
trip_data_july_2020 <- mutate(trip_data_july_2020,start_station_id = as.character(start_station_id),end_station_id = as.character(end_station_id))
trip_data_aug_2020 <- mutate(trip_data_aug_2020,start_station_id = as.character(start_station_id),end_station_id = as.character(end_station_id))
trip_data_sept_2020 <- mutate(trip_data_sept_2020,start_station_id = as.character(start_station_id),end_station_id = as.character(end_station_id))
trip_data_oct_2020 <- mutate(trip_data_oct_2020,start_station_id = as.character(start_station_id),end_station_id = as.character(end_station_id))
trip_data_nov_2020 <- mutate(trip_data_nov_2020,start_station_id = as.character(start_station_id),end_station_id = as.character(end_station_id))
```

#### Let's see if the changes have been made

#### Now we will add some new columns to the merged dataset based on calculations of already existing columns in order to facilitate our analysis and arrive at more insightful conclusions.

```r
# Adding new column to calculate the ride length in minutes
trip_data_all_cleaned$ride_length <- difftime(trip_data_all_cleaned$ended_at, trip_data_all_cleaned$started_at, units="mins")

# Adding new columns to find  month & day of the week of the rides
trip_data_all_cleaned$month <- month(trip_data_all_cleaned$started_at, label=TRUE, abbr=FALSE)
trip_data_all_cleaned$day_of_week <- wday(trip_data_all_cleaned$started_at, label=TRUE, abbr=FALSE)
trip_data_all_cleaned$date <- as.Date(trip_data_all_cleaned$started_at)
```

```r
head(trip_data_all_cleaned)
```

```r
# Since ride_length is in factor type, we will convert it to numeric type for easier calculations
trip_data_all_cleaned$ride_length <- as.numeric(trip_data_all_cleaned$ride_length)
```

```r
# Checking if data type of ride_length is numeric
is.numeric(trip_data_all_cleaned$ride_length)
```

#### I noticed that some data in "ride_length" column took more than 1 day and also some negative values which is not possible since ride length can't be negative, so it's best to remove them from our dataset to conduct accurate analysis 

```r
# Removing rows where ride_length > 1 day (1440 minutes) and < 0 minutes
trip_data_all_cleaned <- trip_data_all_cleaned[!(trip_data_all_cleaned$ride_length > 1440 | trip_data_all_cleaned$ride_length <= 0),]
```

#### PHASE 4 - ANALYZE

#### 1. Conducting Descriptive Analysis :

```r
# Calculating minimum , maximum, mean & median ride_length
trip_data_all_cleaned %>% 
  summarise(min_ride_length = min(ride_length), 
            max_ride_length = max(ride_length),
            median_ride_length = median(ride_length),
            mean_ride_length = mean(ride_length))
```

```r
# Finding number of users in each customer category
trip_data_all_cleaned %>% group_by(member_casual) %>% count()
```

```r
# Compare the ride length between members and casual riders
aggregate(trip_data_all_cleaned$ride_length ~ trip_data_all_cleaned$member_casual, FUN = mean)
aggregate(trip_data_all_cleaned$ride_length ~ trip_data_all_cleaned$member_casual, FUN = median)
aggregate(trip_data_all_cleaned$ride_length ~ trip_data_all_cleaned$member_casual, FUN = max)
aggregate(trip_data_all_cleaned$ride_length ~ trip_data_all_cleaned$member_casual, FUN = min)
```

#### 2. Comparing Member and Casual users on the basis of days of week

```r
aggregate(trip_data_all_cleaned$ride_length ~ trip_data_all_cleaned$member_casual + trip_data_all_cleaned$day_of_week, FUN = mean)
aggregate(trip_data_all_cleaned$ride_length ~ trip_data_all_cleaned$member_casual + trip_data_all_cleaned$day_of_week, FUN = median)
aggregate(trip_data_all_cleaned$ride_length ~ trip_data_all_cleaned$member_casual + trip_data_all_cleaned$day_of_week, FUN = max)
aggregate(trip_data_all_cleaned$ride_length ~ trip_data_all_cleaned$member_casual + trip_data_all_cleaned$day_of_week, FUN = min)
```

#### 3. Fixing days of the week

```r
trip_data_all_cleaned$day_of_week <- ordered(trip_data_all_cleaned$day_of_week, levels=c("Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"))
```

#### 4. Analyzing ridership data by type and day_of_week

```r
trip_data_all_cleaned_v2 <- trip_data_all_cleaned %>% 
  mutate(day_of_week=wday(started_at, label = TRUE)) %>% 
  group_by(member_casual, day_of_week) %>% 
  summarise(number_of_rides=n(), average_duration=mean(ride_length)) %>% 
  arrange(member_casual, day_of_week)
```

#### 5. Analyzing ridership data by type and month

```r
trip_data_all_cleaned_v3 <- trip_data_all_cleaned %>% 
  mutate(month=month(started_at, label = TRUE)) %>% 
  group_by(member_casual, month) %>% 
  summarise(number_of_rides=n(), average_duration=mean(ride_length)) %>% 
  arrange(member_casual, month)
```

#### 6. Inspecting the newly formed dataframes

```r
View(trip_data_all_cleaned_v2)
View(trip_data_all_cleaned_v3)
```

#### 7. Analyzing the top 10 most travelled routes for Member and Casual users

##### Top 10 travelled routes for member riders


```r
top10_station_routes_members <- trip_data_all_cleaned %>% 
  unite("travelled_route", start_station_name, end_station_name, sep = " -- ")

mtr_members <- sqldf("SELECT travelled_route, COUNT(member_casual) AS number_of_member_riders
      FROM top10_station_routes_members
      WHERE member_casual = 'member'
      GROUP BY travelled_route
      ORDER BY number_of_member_riders DESC
      LIMIT 10")

View(mtr_members)
```

```r
mtr_casual_riders <- sqldf("SELECT travelled_route, COUNT(member_casual) AS number_of_casual_riders
      FROM top10_station_routes_members
      WHERE member_casual = 'casual'
      GROUP BY travelled_route
      ORDER BY number_of_casual_riders DESC
      LIMIT 10")

View(mtr_casual_riders)
```

### Summary of the analysis

* The number of rides taken by member users in a week is **3x** greater than the number of rides taken by casual users during the same time period.

* The average ride duration for a casual user in a week is about **4x** the average ride duration for a member user.

* Around **44%** of casual users use the Cyclistic bike sharing services on weekends alone (Sat and Sun).

* Around **80%** of member users use the cyclistic bike sharing services on weekdays (Mon to Fri)

* Average ride duration for member users is constant for most of the week is **15 mins** except on weekends where it is slightly greater at about **16 mins** this suggests that riders are work employees and people who go for short errands. 

* Average ride duration for casual users is almost consant for the week at about **35 minutes**. This suggests the riders are tourist people who go for leisure activities which is explained by long trip duration.

* The top 10 most travelled routes for both member and casual users does not contain any common route. This suggests that the routes taken by member and casual users are very different from each other.

* The highest number of rides for both member and casual users occur between the months of **July to September.**

* Enhance chances to convert casual customers into annual customers.
* Use seasonal campaigns to increase margins.
* Increase visibility among your customers.
